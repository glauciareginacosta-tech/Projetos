## Sistema de Dicas de Investimentos geradas por IA

### Busca Informações e acionamentos
Esta parte do código busca a informações dos usuários em um site e aciona IA para cada um desses usuários para criar dicas de investimentos a partir do seu perfil de investidor e  saldo, posteriormente gera um arquivo com todas essas informações.


`Critério Aceite` - Arquivo com as informações do usuário e dica de investimento gerada no IA criado com sucesso.


```python
import requests
import pandas as pd

URL = ("https://digitalinnovationone.github.io/dio-lab-assistente-investimentos-rpa-n8n/")
res = requests.get(URL)
HTML = res.content

df = pd.read_html(HTML)[0]
#df.shape

# Lista para armazenar as dicas de investimento
investment_tips = []

for i in range(len(df)):
    # Captura a dica de investimento gerada pela função DicaInvest
    current_tip = DicaInvest(df.Nome[i], df.Saldo[i], df.Perfil[i])

    # Adiciona a dica à lista
    investment_tips.append(current_tip)

# Adiciona as dicas de investimento como uma nova coluna ao DataFrame original
df['Dica de Investimento'] = investment_tips

# Define o nome do arquivo CSV de saída
output_csv_filename = 'investment_data_with_tips.csv'

# Salva o DataFrame combinado em um arquivo CSV.
# O parâmetro index=False evita que o índice do DataFrame seja gravado como uma coluna no CSV.
df.to_csv(output_csv_filename, index=False)
print(f"Os dados do HTML e as dicas de investimento foram salvos em {output_csv_filename}")

display(df)
```

###IA
Código de acionamento da IA para dar dicas de investimentos aos usuários.

`Modelo IA` : Nex AGI: Nex-N2.5-Mini (free) - pelo OpenRouter.


```python
import os
from google.colab import userdata

def DicaInvest(nome, saldo, perfil):

    from openrouter import OpenRouter

    # Retrieve API key from environment variable
    api_key_value = userdata.get("OPENROUTER_API_KEY")

    if not api_key_value:
        raise ValueError("OPENROUTER_API_KEY environment variable not set.")

    with OpenRouter(api_key=api_key_value) as client:
        response = client.chat.send(
            model="nex-agi/nex-n2.5-mini:free",
            messages=[
                {
                    "role": "user",
                    "content": f"Você é um excelente consultor de investimento, dê uma dica personalizada de investimento em ações, fundos ou renda fixa para a pessoa " + str(nome)+ " que possui esse valor investido " + str(saldo) + ", e o seu perfil de investidor é " + str(perfil) + ".  Sua dica tem que ser suscinta, formal, técnica mas com linguagem que não entendedores de investimento entenda, resumida e rápida, com no máximo 200 caracteres. você não deve utilizar na dica nem o perfil nem o saldo da usuário. Entretanto em todas as dicas, diga o nome do usuário para ficar bastante pessoal."
                }
            ],
        )

        #Retorno do resultado da IA generativa
        return response.choices[0].message.content
```
