# Development

```python
from crewai import Agent, Task, Crew, Process
import os
from google.colab import userdata
from  datetime import datetime
os.environ['OPENROUTER_API_KEY'] = userdata.get('OPENROUTER_API_KEY')

from crewai import LLM

hoje = datetime.now().date()
arquivo = f'ideias_{hoje}.md'

# The system is utilizing OpenRouter to enable CrewAI to recognize the API."
llm = LLM(
    model="openrouter/nvidia/nemotron-3-ultra-550b-a55b:free",
    base_url="https://openrouter.ai/api/v1",
    api_key=userdata.get('OPENROUTER_API_KEY')
)

#agent
agente_rascunho = Agent(
    role='Criador de Rascunho',
    goal='Escrever ideias iniciais para textos criativos',
    backstory='Escritor ágil especializado em criar listas e ideias rapidamente',
    llm=llm
)

#task
tarefa_GerarIdeias = Task(
    description='criar 3 idéias curtas para posts sobre os benefícios de se estudar o machinelearning',
    expected_output='3 bullets points cada um com uma idéia curta e criativa. O tamanho do bullet deve ser menor que 400 caracteres',
    agent=agente_rascunho,
    output_file=arquivo
)
#crew
equipe = Crew (
    agents=[agente_rascunho],
    tasks=[tarefa_GerarIdeias],
    process=Process.sequential
)

#result
resultado = await equipe.kickoff_async()
print(resultado)
```
