* **2 – AI Justification Function**

This function was created for generative AI to provide a justification for each counting error it made when calculating function points.

For the query, the free model Tencent: Hy3 (free) was used. My key expires in 7 days (as of 07/09/2026).

Observation: This code, in which the artificial intelligence justifies a counting error, was conceived as a playful exercise to enable it to transparently acknowledge its mistakes. It is evident that there is no practical purpose in this; it is merely a way to include artificial intelligence in this pipeline. After all, the models used to perform the APF counting are entirely different from those employed to justify the errors, and no analysis of the prompts was conducted.

```python
import os
from google.colab import userdata

def Funcao_PFIA(TamanhoReal, TamanhoIA):

    from openrouter import OpenRouter

    # Retrieve API key from environment variable
    api_key_value = userdata.get("OPENROUTER_API_KEY")

    if not api_key_value:
        raise ValueError("OPENROUTER_API_KEY environment variable not set.")

    with OpenRouter(api_key=api_key_value) as client:
        response = client.chat.send(
            model="tencent/hy3:free",
            messages=[
                {
                    "role": "user",
                    "content": f"Você achou que sabia ponto de função e o seu cálculo deu " + str(TamanhoReal) + ", entretanto o tamanho funcional correto é" + str(TamanhoIA) + ". Faça uma justificativa para esse seu equívoco. Sua justificativa tem que ser suscinta, resumida e rápida, com no máximo 50 caracteres."
                }
            ],
        )

        #Retorno do resultado da IA generativa
        return response.choices[0].message.content
```

➤➤➤➤ **IN ENGLISH**

**3 - Transformation -> Step 1**

Calculation of the error percentage and the variance between the actual functional sizes and those calculated by artificial intelligence for each demand;

Inclusion of the AI's justification for discrepancies in the sizing of each demand.

```python
TamanhoFuncional = pd.DataFrame(TamanhoFuncional)

#Adicionar o tamanho em variáveis
PF_IA = (TamanhoFuncional['PF_IA'])
PF = (TamanhoFuncional['PF'])

#Percentual do erro
TamanhoFuncional['%Erro'] = ((PF_IA - PF) / PF) * 100
#Diferença
TamanhoFuncional['Diferenca'] = PF_IA - PF

#Justificativa
# Chama a função Funcao_PFIA (que usa IA para justificar a contagem) para cada linha do dataframe para atualizar a coluna Justificativa.
TamanhoFuncional['Justificativa'] = TamanhoFuncional.apply(lambda row: Funcao_PFIA(row['PF'], row['PF_IA']), axis=1)

TamanhoFuncional.tail()
```

**3 - Transformação -> Passo 2**

Cálculo subsequente das métricas de mínimo, máximo, média, mediana e desvio-padrão para o tamanho funcional real e para o gerado pela IA.




➤➤➤➤ **IN ENGLISH**

**3 - Transformation -> Step 2**

Subsequent calculation of the minimum, maximum, mean, median, and standard deviation metrics for both the actual and AI-generated functional sizes;

```python
from numpy import sqrt

#Mínimo
TamanhoFuncional['Minimo_PF'] = min(TamanhoFuncional['PF'])
TamanhoFuncional['Minimo_PF_IA'] = min(TamanhoFuncional['PF_IA'])
#Máximo
TamanhoFuncional['Maximo_PF'] = max(TamanhoFuncional['PF'])
TamanhoFuncional['Maximo_PF_IA'] = max(TamanhoFuncional['PF_IA'])
#Média
TamanhoFuncional['Media_PF'] = TamanhoFuncional['PF'].mean()
TamanhoFuncional['Media_PF_IA'] = TamanhoFuncional['PF_IA'].mean()
#Mediana
TamanhoFuncional['Mediana_PF'] = TamanhoFuncional['PF'].median()
TamanhoFuncional['Mediana_PF_IA'] = TamanhoFuncional['PF_IA'].median()
#Desvio padrão
TamanhoFuncional['Desvio_PF'] = TamanhoFuncional['PF'].std()
TamanhoFuncional['Desvio_PF_IA'] = TamanhoFuncional['PF_IA'].std()
```

```python
print(TamanhoFuncional)
```

**3 - Transformação -> Passo 3**

Cálculo final do total geral do percentual de erro e da diferença acumulada.



➤➤➤➤ **IN ENGLISH**

**3 - Transformation -> Step 3**

Final consolidation of the overall error percentage total and the accumulated variance.

```python
#Somas dos tamanhos funcionais (reais e calculados por IA)
SOMA_PF_IA = (TamanhoFuncional['PF_IA']).sum()
SOMA_PF = (TamanhoFuncional['PF']).sum()

#Percentual do erro - total
TamanhoFuncional['%Erro_Total'] = ((SOMA_PF_IA - SOMA_PF) / SOMA_PF) * 100
#Diferença - total
TamanhoFuncional['Diferenca_Total'] = SOMA_PF_IA - SOMA_PF
```
